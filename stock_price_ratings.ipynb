## Stock Price and Analyst Ratings Automation. 
For a given stock ticker (get_ticker_given_name)
1. Latest stock price (get_stock_price)
2. Analyst Ratings (get_analyst_rating_summary)

```python
import yfinance as yf

def get_price(ticker):
    stock = yf.Ticker(ticker)
    
    price = stock.info.get("currentPrice")
    
    print(f"{ticker} current price: {price}")
```

```python
get_price("MU")
```

```python
def get_analyst_ratings(ticker):
    stock = yf.Ticker(ticker)

    # raw analyst recommendation history / table
    recs = stock.recommendations

    if recs is None or len(recs) == 0:
        print(f"No analyst recommendations found for {ticker}")
        return

    print(f"\nAnalyst recommendations for {ticker}:\n")
    print(recs.tail(10))   # show latest 10 rows
```

```python
get_analyst_ratings("MU")
```
